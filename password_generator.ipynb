# Simple Password generator written by JMW

A very simplistic password generator I am already using for a while now.
I wanted to share it with the rest of the world.

The idea is rather simple, a password is generatred by selecting a given number of characters from one of 3 given character sets.
I like to seperate between "base characters", "numbers" and "special symbols". But this is puerly optional.
The final sequence is generated via randomly shuffling the selected characters.


```python
# IMPORT:

import numpy as np
import time
import psutil
from datetime import datetime

print("success")
%pwd 
```

```python
# PARAMETERS

characters_str_1 = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ" # first set of characters
characters_str_2 = "0123456789" # second set of characters
characters_str_3 = "!?.#+<>:" # third set of characters

number_chars_1 = 10 # number of chars from set 1
number_chars_2 = 4 # number of chars from set 2
number_chars_3 = 4 # number of chars from set 3

warmup = 11235 # random number of your choice
```

```python
# generate random password


# generate random seed
seed = time.time() - psutil.boot_time()

# initiate rng gen
rng = np.random.default_rng(seed=int(seed)) 

# warump rng
for i in range(warmup): rng.random()

# generate character arrays from given strings
chars_1 = list(characters_str_1)
chars_2 = list(characters_str_2)
chars_3 = list(characters_str_3)

# shuffle sets
rng.shuffle(chars_1)
rng.shuffle(chars_2)
rng.shuffle(chars_3)

# randomly select characters from arrays
selec_1 = rng.choice(chars_1, number_chars_1)
selec_2 = rng.choice(chars_2, number_chars_2)
selec_3 = rng.choice(chars_3, number_chars_3)

# merge character arrays to one array
selec_concat = np.concatenate([selec_1, selec_2, selec_3], axis=0)

# shuffle final array
rng.shuffle(selec_concat)

# make final string
seq = ''.join(selec_concat) 

# print final reult
print(f"FINAL SEQUENCE: {seq}")
print(seq)

```
